## **Core Logging Exercsise with stratapy**

stratapy is a Python package which was made by PhD students from the school for visualising stratigraphic logs in a digital and consistent format.

The tool reads in Excel or ``csv`` files to produce a graphical stratigraphic log which can be saved as an image file.

stratapy offers a wide range of customisation option, including log format (e.g. log, grainsize axis, lithology column), lithology patterns and colours, palaeontological and stratigraphic symbols, and much more.

**This exercise will guide you through creating a stratigraphic log using stratapy and then let you explore the package to create your own log using the core samples available**.

Our lithological patterns are taken from the United States Geological Survey (USGS) pattern library, and you can view all of the available patterns in the [stratapy documentation](https://stratapy.readthedocs.io/en/latest/reference/available_assets.html), or in the image provided with this notebook. ##Customising Lithology Patterns, Colours and Names

To edit the lithology patterns, colours and names, we can use the ``update_lithologies()`` function. This function takes a dictionary as an argument.

First, we define a new 'key' like ``my_sandstone``. Then, we choose three values for this key:
- the name or numerical key of a default lithology pattern to use in this new key, e.g. ``'sandstone'``
- the colour to use for this new key, e.g. ``'green'`` or a hex code like ``'#00FF00'``
- the name to use for this new key, e.g. ``'My Sandstone'

We can use as many of these new keys as we want, and they will be added to the existing lithology patterns.

To use them, simply use the new key in the ``rock`` column of your ``.csv`` file, and stratapy will use the new pattern, colour and name for that unit. 

The code cell below shows an example of how to use the ``update_lithologies()`` function to add new lithology patterns, colours and names.

````python
#for e.g.,
sp.update_lithologies({
    'new_667':  ('667', '', 'Anhydrite'),
    'new_669':  ('669', '', 'Interbedded sands, silts, coals'),
    'new_608':  ('608', "#676565FF", 'Laminated sands'),
    'new_608a': ('608', "#323131FF", 'Laminated sands'),
})
````
