This notebook simply runs stratapy on the ``.csv`` files in the ``answers`` folder which represent ideally digitised logs.

The ``.csv`` files in this /answers folder have grain size values and entries for palaeontological and stratigraphic features. You can choose to display them as a standard core log with ``display_mode='log'`` or with a grain size axis using ``display_mode='grain_size'`` or ``display_mode='default'``.

```python
# Ensure stratapy is installed
# !pip install stratapy

# Import stratapy
import stratapy as sp
```

```python
log = sp.load('54-1b-2.csv')
log.plot(figsize=(3, 12), display_mode='log', y_label='Depth (ft)')
log.save('54-1b-2.png')
```

```python
log = sp.load('49-22-55-upper.csv')
log.plot(figsize=(3, 12), display_mode='log', y_label='Depth (ft)')
log.save('49-22-55-upper.png')
```

```python
log = sp.load('49-22-55-lower.csv')
log.plot(figsize=(3, 12), display_mode='log', y_label='Depth (ft)')
log.save('49-22-55-lower.png')
```
